```python
path = "C:\\Users\\Admin\\OneDrive\\Desktop\\UniBE\\data\\150525_530_15_1a_DRC470_02.brw"
rec = si.BiocamRecordingExtractor(path)
print(rec)
```

```python
sx = si.run_sorter(
    "herdingspikes",    
    rec,
    folder="hs_output",
    verbose=True,
    remove_existing_folder=True,
)
se.NpzSortingExtractor.write_sorting(sx, 'sorting.npz')
```

```python
# wrap your Biocam recording before filtering
rec_filled = spre.FillMissingValuesRecording(rec, mode='zeros')  # or mode='synthetic_noise'

# now continue with scaling and filtering
rec_scaled = spre.scale(rec_filled)
rec_filtered = spre.bandpass_filter(rec_scaled, freq_min=300, freq_max=4500)
```

```python
params = {
    "fill_gaps_strategy": "zeros",  # still good to include in case of missing frames
    "save_shape": False  # Set to True if you want to store spike shapes
}

det = hs.HSDetectionLightning(rec_filtered, params)
det.DetectFromRaw()
det.PlotDensity(binsize=4)
```
